```python
import jax
import matplotlib.pyplot as plt
key = jax.random.PRNGKey(1234)
key, sk = jax.random.split(key)

#used in Master thesis
v_tr_uni = jax.random.uniform(sk, (100000, ))*5900 + 100 
min(v_tr_uni), max(v_tr_uni)

```

```python
#Does not work 
v_tr_gamma1 = jax.random.gamma(sk, 3.0, (100000, ))*300*1
min(v_tr_gamma1), max(v_tr_gamma1)

```

```python
#Does not work 

v_tr_gamma2 = jax.random.gamma(sk, 3.0, (100000, ))*300*2
min(v_tr_gamma2), max(v_tr_gamma2)

```

```python
#Does work 

v_tr_gamma12 = jax.random.gamma(sk, 3.0, (100000, ))*300*1.2
min(v_tr_gamma12), max(v_tr_gamma12)
```

```python
fig, ax = plt.subplots(nrows = 4, sharex = True , sharey = True)
ax[0].hist(v_tr_uni, bins = 50)
ax[1].hist(v_tr_gamma1, bins = 50)
ax[2].hist(v_tr_gamma2, bins = 50)
ax[3].hist(v_tr_gamma12, bins = 50)
plt.show()
```
